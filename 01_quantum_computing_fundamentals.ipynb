## What Is Quantum Computing?

A classical computer works with bits. A bit is always either `0` or `1`.

A quantum computer works with qubits. A qubit can exist in a superposition, which means it can behave like a combination of `0` and `1` before measurement.

### A simple analogy

- Classical bit: a coin lying flat on a table
- Quantum bit: a coin spinning in the air

While the coin is spinning, it is not simply heads or tails. In the same way, a qubit is not forced into a single state until we measure it.

This is one of the core ideas that makes quantum computing different from classical computing.

## What Is a Quantum Circuit?

If you have built software before, you already understand workflows.

A typical program does something like this:

1. Accept input
2. Process the data
3. Return a result

A quantum circuit follows a similar idea, but it processes qubits instead of regular variables.

A quantum circuit is simply a sequence of instructions applied to qubits.

### The basic flow

- Create qubits
- Apply quantum gates
- Measure the result
- Read the output

At a high level, that is the foundation of quantum programming.


## Quantum Gates Explained Like a Python Developer

If you have written Python before, you are already used to changing values.

```python

```

This flips a value from `False` to `True`.

A quantum gate is similar in spirit: it is a special operation that changes the state of a qubit.

### Common quantum gates

- `X` gate
- `H` gate
- `CX` (CNOT) gate

Each of these gates does something different, and together they form the basic vocabulary of quantum circuits.


````python

from qiskit import QuantumCircuit

# Create a circuit with 2 qubits and 2 classical bits
qc = QuantumCircuit(2, 2)

# Apply the Hadamard gate to qubit 0
qc.h(0)

# Display the circuit
print(qc)
````

## X Gate: The Quantum Light Switch

Imagine a room light switch.

- If the light is off, pressing the switch turns it on.
- If the light is on, pressing the switch turns it off.

That is the classical idea behind a flip operation.

The `X` gate does the same job for a qubit:

- `0` becomes `1`
- `1` becomes `0`

This is the quantum equivalent of a toggle switch, and it is often described as the quantum version of a logical NOT.


````python
qc.x(0)
````

## H Gate: The Spinning Coin Trick

The `H` gate, or Hadamard gate, is one of the most important gates in quantum computing.

In classical computing, a bit is definitely either `0` or `1`.

In quantum computing, a qubit can be put into a superposition where both outcomes are possible until measurement.

### Intuition

Think of a coin:

- Before flipping, it has a definite state in your hand.
- While spinning, it is not clearly heads or tails.

The Hadamard gate creates that same kind of “both possibilities at once” behavior.

This is how quantum systems begin to explore multiple states at the same time.


## CX Gate: Making Two Qubits Work Together

The `CX` gate, also called the CNOT gate, is a two-qubit gate.

It works like this:

- One qubit acts as the control.
- The other qubit acts as the target.

If the control qubit is `1`, the target qubit flips. If the control qubit is `0`, nothing happens.

### A real-world analogy

Imagine a manager giving instructions to a teammate:

- If the manager gives the signal, the teammate changes state.
- If no signal is given, nothing changes.

This conditional behavior is what makes two-qubit gates so powerful.

When combined with the Hadamard gate, the CNOT gate can create entanglement.


````python
qc.h(0)
````

## Building Your First Quantum Circuit

Now it is time to build a real circuit in Qiskit.

The first step is to import the needed library and create a small quantum circuit.





````python
from qiskit import QuantumCircuit

qc = QuantumCircuit(2, 2)
print(qc)
````

This creates:

- two qubits
- two classical bits

Right now the circuit is empty, but we are about to add the quantum operations.


## Creating Superposition

The first operation we add is a Hadamard gate:

```python
qc.h(0)
```

This places the first qubit in superposition.

At this moment, the qubit is no longer locked to a single value. It has a probability of being measured as either `0` or `1`.

This is the beginning of quantum behavior.


````python
qc.cx(0, 1)
````

## Creating Entanglement

To create a more interesting quantum state, we add a controlled operation:






````python
qc.cx(0, 1)
````

This links the second qubit to the first one.

The result is a shared quantum state known as entanglement. Once the qubits become entangled, their measurement results are correlated.

In a Bell state, the qubits always agree: either both are `0` or both are `1`.

## Measuring the Qubits

Quantum states are not directly readable by a computer, so we must measure them.

```python
qc.measure([0, 1], [0, 1])
```

This tells Qiskit to:

- measure qubit `0` and store it in classical bit `0`
- measure qubit `1` and store it in classical bit `1`

At this point, our circuit is complete and ready to run.


````python

from qiskit import QuantumCircuit

# Create a circuit with 2 qubits and 2 classical bits
qc = QuantumCircuit(2, 2)

# Display the circuit
print(qc)
````

## Running the Circuit on a Simulator

You do not need a real quantum computer to learn this.

Qiskit provides a local simulator called `AerSimulator`.






````python

from qiskit_aer import AerSimulator

simulator = AerSimulator()
qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
qc.measure([0, 1], [0, 1])

result = simulator.run(qc, shots=1024).result()
counts = result.get_counts()
print(counts)
````

### Why use shots?

A single run is not enough to reveal the pattern, so we run the circuit many times.

`shots=1024` means the circuit is executed 1,024 times, and the output is summarized as counts.

This is similar to flipping a coin many times to understand the probability distribution.

## Your Complete Bell State Program

This is the full version of the program we just built:






````python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
qc.measure([0, 1], [0, 1])

simulator = AerSimulator()
result = simulator.run(qc, shots=1024).result()
counts = result.get_counts()
print(counts)
````

The result is usually something like:

```python
{'00': 500, '11': 524}
```

You should almost never see `01` or `10`.

That is the sign that the qubits are entangled correctly.


````python
from qiskit import QuantumCircuit

qc = QuantumCircuit(2, 2)

qc.h(0)

print(qc)
````

## Visualizing the Results

Developers often prefer visual feedback. A histogram makes the pattern easier to see.





````python
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

plot_histogram(counts)
plt.show()
````

When the circuit works correctly, the histogram shows a strong preference for:

- `00`
- `11`

The result is random from run to run, but the correlation remains consistent.

## Understanding What Just Happened

This is the big idea.

We started with two qubits, created superposition on the first qubit, entangled them with a CNOT gate, and then measured them.

The result was remarkable:

- both qubits were either `00` or `11`
- never `01` or `10`

This reveals entanglement, one of the most important ideas in quantum mechanics.

### Bell state in one sentence

A Bell state is a maximally entangled two-qubit state that shows the strongest form of correlation possible in quantum physics.
